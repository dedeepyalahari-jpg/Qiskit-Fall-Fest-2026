```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()

# Part A: Correlated outcomes
qc_a = QuantumCircuit(2, 2)
qc_a.h(0)
qc_a.cx(0, 1)
qc_a.measure([0, 1], [0, 1])

qc_a.draw("mpl")
```

```python
result_a = simulator.run(qc_a, shots=100).result()
counts_a = result_a.get_counts()

print("Part A results:", counts_a)
plot_histogram(counts_a)
plt.show()
```

```python
# Part B: Opposite outcomes
qc_b = QuantumCircuit(2, 2)
qc_b.h(0)
qc_b.cx(0, 1)
qc_b.x(0)
qc_b.measure([0, 1], [0, 1])

qc_b.draw("mpl")
```

```python
result_b = simulator.run(qc_b, shots=100).result()
counts_b = result_b.get_counts()

print("Part B results:", counts_b)
plot_histogram(counts_b)
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['00', '11', '01', '10']
values = [counts_a.get('00', 0),
          counts_a.get('11', 0),
          counts_b.get('01', 0),
          counts_b.get('10', 0)]

plt.figure(figsize=(8, 5))
plt.bar(labels, values)
plt.xlabel("Measurement Outcome")
plt.ylabel("Number of Shots")
plt.title("Quantum Correlation Switch")
plt.show()
```

Challenge 4: Quantum Correlation Switch

Part A: Correlated Outcomes

- Circuit: Hadamard gate followed by CNOT gate.
- Prediction: Outcomes 00 and 11 will dominate.
- Actual results: 00 = 53, 11 = 47.

Part B: Opposite Outcomes

- Circuit: Hadamard gate, CNOT gate, and X gate on qubit 0.
- Prediction: Outcomes 01 and 10 will dominate.
- Actual results: 01 = 45, 10 = 55.

Comparison

In Part A, the two qubits produce correlated outcomes (00 and 11). In Part B, applying the X gate to qubit 0 changes the results to opposite outcomes (01 and 10).

Conclusion

Quantum gates can change the relationship between the measurement outcomes of two qubits. The histograms show the difference between correlated and opposite outcomes using 100 shots for each circuit.
