```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt

simulator = AerSimulator()

# Circuit A: Apply one X gate
qc_a = QuantumCircuit(1, 1)
qc_a.x(0)
qc_a.measure(0, 0)

# Circuit B: Apply two X gates
qc_b = QuantumCircuit(1, 1)
qc_b.x(0)
qc_b.x(0)
qc_b.measure(0, 0)

# Circuit C: Apply one H gate
qc_c = QuantumCircuit(1, 1)
qc_c.h(0)
qc_c.measure(0, 0)

circuits = [qc_a, qc_b, qc_c]
names = ["Circuit A: X", "Circuit B: X + X", "Circuit C: H"]

for name, circuit in zip(names, circuits):
    print(name)
    display(circuit.draw("mpl"))

    result = simulator.run(
        circuit, shots=100
    ).result()

    counts = result.get_counts()
    print("Measurement results:", counts)

    plt.figure()
    plt.bar(counts.keys(), counts.values())
    plt.xlabel("Measurement outcome")
    plt.ylabel("Number of shots")
    plt.title(name)
    plt.show()

```

Challenge 1: Predict the Output

1. Circuit A — Single X Gate

Prediction: The output will be 1 for all 100 shots.

Actual result: "{'1': 100}"

Observation: The X gate changes the qubit state from |0⟩ to |1⟩. The prediction matches the actual result.

2. Circuit B — Two X Gates

Prediction: The output will be 0 for all 100 shots.

Actual result: "{'0': 100}"

Observation: Two consecutive X gates return the qubit to its initial state |0⟩. The prediction matches the actual result.

3. Circuit C — Single Hadamard Gate

Prediction: Approximately 50% of measurements will be 0 and 50% will be 1.

Actual result: "{'0': 49, '1': 51}"

Observation: The Hadamard gate creates an equal superposition of |0⟩ and |1⟩. The measured counts are close to the predicted distribution.

Conclusion

The experiment demonstrates how quantum gates affect a qubit. One X gate flips the state, two X gates cancel each other, and the Hadamard gate produces a superposition. The experimental results agree with the theoretical predictions.
