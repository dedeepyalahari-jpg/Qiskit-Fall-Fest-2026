```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt

simulator = AerSimulator()

# Circuit A: Make 0 occur more frequently
qc_a = QuantumCircuit(1, 1)
qc_a.ry(0.4, 0)
qc_a.measure(0, 0)

# Circuit B: Make 1 occur more frequently
qc_b = QuantumCircuit(1, 1)
qc_b.ry(2.4, 0)
qc_b.measure(0, 0)

circuits = [qc_a, qc_b]
names = ["Circuit A: 0 more frequent", "Circuit B: 1 more frequent"]

for name, circuit in zip(names, circuits):
    print(name)
    display(circuit.draw("mpl"))

    result = simulator.run(circuit, shots=100).result()
    counts = result.get_counts()

    print("Measurement results:", counts)

    plt.figure()
    plt.bar(counts.keys(), counts.values())
    plt.xlabel("Measurement outcome")
    plt.ylabel("Number of shots")
    plt.title(name)
    plt.show()
```

Challenge 2: Make the Quantum Coin Biased

Circuit A — 0 More Frequent

- Gate used: Ry(0.4)
- Prediction: Outcome 0 will occur more frequently.
- Actual result: 0 = 95, 1 = 5.

Circuit B — 1 More Frequent

- Gate used: Ry(2.4)
- Prediction: Outcome 1 will occur more frequently.
- Actual result: 0 = 23, 1 = 77.

Conclusion

The Ry gate changes the quantum state and the probabilities of measuring 0 or 1. Circuit A produced outcome 0 in 95% of shots, while Circuit B produced outcome 1 in 77% of shots.
